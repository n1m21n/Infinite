# M1 forward model (Colab)
SOUND-MODEL.md milestone M1: knobs + note + velocity -> descriptors and log-mel, PyTorch. Any runtime works (CPU about 10 min at 3,000 presets; a GPU is faster).

Rows and code come from Drive (`Infinite-Cloud-Training`, refreshed from the Mac with `python3 -I tools/universe/sync_drive.py`). Only feature rows travel, no audio. Results go back to Drive as `models/m1_<rows>_colab`.

```
from google.colab import drive
drive.mount('/content/drive')
```

```
import os, gzip, shutil, glob
DRIVE = '/content/drive/MyDrive/Infinite-Cloud-Training'
HOME = '/content/home'; os.environ['HOME'] = HOME
TD = HOME + '/Infinite/training-data'; UN = HOME + '/Infinite/repo/tools/universe'
ROWS = 'audio/preset_diva_v2.jsonl'          # any row-v2 file
os.makedirs(TD + '/audio', exist_ok=True); os.makedirs(UN, exist_ok=True)
for f in glob.glob(DRIVE + '/universe/*.py'):
    shutil.copy(f, UN)
src = DRIVE + '/training-data/' + ROWS
if os.path.exists(src + '.gz'):
    with gzip.open(src + '.gz', 'rb') as a, open(TD + '/' + ROWS, 'wb') as b: shutil.copyfileobj(a, b)
else:
    shutil.copy(src, TD + '/' + ROWS)
print(os.path.getsize(TD + '/' + ROWS) >> 20, 'MB of rows;', len(os.listdir(UN)), 'tools')
```

## Train and score
Prints baselines (training mean, k-NN), each seed, the 3-seed average and the M1 bars as PASS / FAIL.

## Write results back to Drive

```
name = os.path.basename(ROWS).split('.')[0]
shutil.copytree(f'{TD}/models/m1_{name}', f'{DRIVE}/training-data/models/m1_{name}_colab', dirs_exist_ok=True)
print(open(f'{TD}/models/m1_{name}/results.txt').read()[-3000:])
```
